```python
# Bridge SA · Operations — altas de la cohorte de marzo
#
# ALTAS_TEXTO ya esta cargado con el listado, tal cual.
# Cada linea util tiene la forma:  Nombre ; correo ; programa
#
# Deja preparadas estas tres cosas:
#   altas_limpias -> lista de listas [nombre, email, programa]
#   descartes     -> lista de listas [linea_original, motivo]
#   artefacto     -> el paquete que se lleva el Sprint 2
#
# Recuerda: linea en blanco se salta (no es descarte); correo sin @ o que
# no acabe en @bridge.example es descarte "email_invalido"; correo ya
# visto es descarte "duplicado". Y primero se limpia, despues se decide.
#
# Trucos: " ".join(t.split()) deja un solo espacio entre palabras;
#         "".join(t.split()) los quita todos;  t.title() pone la inicial
#         de cada palabra en mayuscula.


# LISTAS
altas_limpias = []
descartes = []
vistos = []

# ANALIZAR LISTADO ORIGINAL
for linea in ALTAS_TEXTO.split("\n"):

    # TODO: saltar líneas en blanco
    if linea.strip() == "":
        continue

    print(linea) # IMPRIIMR LINEA 
    campos = linea.split(";") # SEPARAR LINEA EN COLUMNAS

    # COMPROBAR Y ASIGNAR COLUMNAS
    if len(campos) == 3:
        print("Campos correctos")

        # DEFINIR COLUMNAS
        nombre = campos[0] 
        email = campos[1]
        programa = campos [2]

        print (nombre)
        print (email)
        print (programa)

        # LIMPIAR Y NORMALIZAR DATOS
        nombre = " ".join(nombre.split()).title()
        email = "".join(email.split()).lower()
        programa = " ".join(programa.split()).title()

        if programa == "Ia Engineering":
            programa = "IA Engineering"

        #COMPROBAR SI FALTAN DATOS
        if nombre == "" or email == "" or programa == "":
            print("Faltan datos")
            descartes.append([linea, "Campos vacíos"])

        #VALIDAR CORREO ELECTRÓNICO
        elif (
            email.count("@") == 1
            and email.endswith("@bridge.example")
            and email.split("@")[0] != ""
            and not any(caracter.isspace() for caracter in email)
        ):

            #DETECTAR CORREOS DUPLICADOS
            if email in vistos:
                print("Email duplicado")
                descartes.append([linea, "duplicado"])

            #GUARDAR INSCRIPCIONES VÁLIDAS
            else:
                vistos.append(email)
                alta = [nombre, email, programa]
                altas_limpias.append(alta)
                print("Inscripción guardada:", alta)

        #DESCARTAR CORREOS INVÁLIDOS
        else:
            print("Email inválido")
            descartes.append([linea, "email_invalido"])
    
    #DESCARTAR INSCRIPCIONES MAL INSERTADAS
    else:
        print("Inscripción incompleta")
        descartes.append([linea, "columnas_invalidas"])


artefacto = {
    "artefacto": "altas_limpias",
    "fuente": "Listado original de altas de la cohorte de marzo de Bridge SA, cargado en ALTAS_TEXTO",
    "columnas": ["nombre", "email", "programa"],
    "altas": altas_limpias,
    "descartes": descartes,
}

print("Altas validas:", len(altas_limpias))
print("Descartes:", len(descartes))


```
